```python
import pandas as pd
import numpy as np

df = pd.read_csv('../dataset/dataset_bai03.csv')
df
```

```python
df.replace(to_replace="?", value=np.nan, inplace=True)
```

```python
# dropna trên cột 'price'
df.dropna( inplace=True)
```

```python
df
```

```python
numeric_cols = [
    'normalized-losses',
    'bore',
    'stroke',
    'horsepower',
    'peak-rpm',
    'price'
]
```

```python
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric)
```

```python
df1 = df.copy()
```

```python
make_mapping = {
    make: i
    for i, make in enumerate(df1['make'].unique())
}

df1['make'] = df1['make'].map(make_mapping)
```

```python
make_mapping
```

```python
df1['num-of-doors'] = df1['num-of-doors'].replace({
    'two': 1,
    'four': 2
}).astype(int)
```

```python
df1['num-of-doors']
```

```python
df2 = df1.copy()
```

```python
df2.select_dtypes(include='str').columns
```

```python
df2 = pd.get_dummies(df2, dtype=int)
```

```python
df2
```

```python
df3 = df2.copy()
```

```python
df3 = (df3 - df3.mean()) / df3.std()
```
