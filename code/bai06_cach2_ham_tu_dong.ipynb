```python
import pandas as pd
from scipy import stats
```

```python
df = pd.read_csv('../dataset/EDA_automobile.csv')
df
```

```python
def corr(s1, s2) -> bool:
    r = s1.corr(s2)
    return r < -0.3 or r > 0.3
```

```python
def anova(df, cot) -> bool:
    nhom = df[[cot, 'price']].groupby(cot)
    ds_nhom = []
    for ten in df[cot].unique():
        ds_nhom.append(nhom.get_group(ten)['price'])
    f, p = stats.f_oneway(*ds_nhom)
    return p < 0.05
```

```python
def ds(df):
    ket_qua = []

    for cot in df.columns:
        if cot != 'price':
            if df[cot].dtype == 'int64' or df[cot].dtype == 'float64':
                if corr(df[cot], df['price']):
                    ket_qua.append(cot)
                elif df[cot].nunique() <= 10 and anova(df, cot):
                    ket_qua.append(cot)

            elif anova(df, cot):
                ket_qua.append(cot)

    return ket_qua
```

```python
ds(df)
```
