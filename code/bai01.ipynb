Tạo 1 danh sách LIST COMPREHENSION: tạo ra list mới từ list cũ

```python
data = [2,5,9,5,7,8]

[i + 1 for i in data] # i+1 là tăng 1 đơn vị cho tất cả các phần tử trong data
```

QUY TẮC tham số

```python
#có thể truyền nhiều cách khác nhau, cách này dùng tham số mặc định của b  ko truyền
def tinh(a,b=2):
   cong= a + b
   tru = a - b
   return cong, tru
```
