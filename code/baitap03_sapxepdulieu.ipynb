```python
import pandas as pd
import numpy as np
df=pd.read_csv('../dataset/dataset_bai03.csv')
df
```

```python
# Chuyển '?' thành NaN để dropna nhận dạng được
df.replace(to_replace="?", value=np.nan, inplace=True)
```

```python
# dropna trên cột 'price'
df.dropna( axis=0,subset=["price"], inplace=True)
df
```

```python
#Xử lý cột horsepower - Replace bằng mean
mean_horsepower = df["horsepower"].astype(float).mean()
```

```python
df["horsepower"] = df["horsepower"].replace( to_replace=np.nan,value=mean_horsepower)
```

```python
df["normalized-losses"] = df["normalized-losses"].astype(float)
```

```python
mode_losses = df["normalized-losses"].mode()[0]
```

```python
df["normalized-losses"] = df["normalized-losses"].replace(to_replace=np.nan, value=mode_losses)
```

```python
df.dropna(inplace=True)
df
```

```python
print(df.dtypes) # kiểm tra kiểu dữ liệu
```

```python
df["price"] = df["price"].astype(int)

df["horsepower"] = df["horsepower"].astype(float)

df["normalized-losses"] = df["normalized-losses"].astype(int)

df["bore"] = df["bore"].astype(float)

df["stroke"] = df["stroke"].astype(float)

df["peak-rpm"] = df["peak-rpm"].astype(float)
```

```python
# Min-Max bằng công thức

df["length_minmax"] = (
    (df["length"] - df["length"].min()) /
    (df["length"].max() - df["length"].min())
)

df["width_minmax"] = (
    (df["width"] - df["width"].min()) /
    (df["width"].max() - df["width"].min())
)

df["height_minmax"] = (
    (df["height"] - df["height"].min()) /
    (df["height"].max() - df["height"].min())
)

df
```

```python
#min max Theo thư viện
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[["length_MinMaxScaler",
    "width_MinMaxScaler",
    "height_MinMaxScaler"]] = scaler.fit_transform(
    df[["length", "width", "height"]]
)
df
```

```python
# Scale công thức

df["length_scale"] = df["length"] / df["length"].max()
df["width_scale"] = df["width"] / df["width"].max()
df["height_scale"] = df["height"] / df["height"].max()

df
```

```python
# scale theo thư viện

from sklearn.preprocessing import MaxAbsScaler

scaler = MaxAbsScaler()

df[["length_MaxAbsScaler",
    "width_MaxAbsScaler",
    "height_MaxAbsScaler"]] = scaler.fit_transform(
    df[["length", "width", "height"]]
)
df
```

```python
# Z score theo thư viện
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df[["length_StandardScaler",
    "width_StandardScaler",
    "height_StandardScaler"]] = scaler.fit_transform(
    df[["length", "width", "height"]]
)
df
```

```python
# Z-score bằng công thức

df["length_zscore"] = (
    (df["length"] - df["length"].mean()) /
    df["length"].std()
)

df["width_zscore"] = (
    (df["width"] - df["width"].mean()) /
    df["width"].std()
)

df["height_zscore"] = (
    (df["height"] - df["height"].mean()) /
    df["height"].std()
)
df
```
