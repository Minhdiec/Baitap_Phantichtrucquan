```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
df=pd.read_csv('../dataset/Model_Dataset.csv')
```

```python
df
```

```python
from sklearn.linear_model import LinearRegression
```

```python
X3 = df[['curb-weight','engine-size','length','width','horsepower',
         'city-mpg','highway-mpg','wheel-base','bore','drive-wheels']]
X3 = pd.get_dummies(X3)
y = df['price']
m3 = LinearRegression()
m3.fit(X3, y)
```

```python
b0_3 = m3.intercept_
b1_3 = m3.coef_
b1_3
```

```python
X3
```

```python
yhat_m3 = m3.predict(X3)
yhat_m3
# của price
```

```python
import seaborn as sns
```

```python
ax1 = sns.kdeplot(data=df, x="price")
sns.kdeplot(x=yhat_m3, ax=ax1)
```

```python
# bài tập  ['horsepower', 'curb-weight', 'engine-size', 'highway-mpg']
# m9: bậc 3, m10: bậc 6
X9 = df[['horsepower','curb-weight','engine-size','highway-mpg']]
```

```python
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.pipeline import Pipeline

Input9 = [
    ("ss", StandardScaler()),
    ("pf9", PolynomialFeatures(degree=3, include_bias=False)),
    ("m9", LinearRegression())
]

m9 = Pipeline(Input9)
m9.fit(X9, y)
```

```python
m9.predict(np.array([[111,2548,130,27]]))
```

```python
yhat_m9 = m9.predict(X9)
yhat_m9
```

```python
ax1 = sns.kdeplot(data=df, x="price")
sns.kdeplot(x=yhat_m9, ax=ax1)
```

```python
Input10 = [
    ("ss", StandardScaler()),
    ("pf10", PolynomialFeatures(degree=6, include_bias=False)),
    ("m10", LinearRegression())
]

m10 = Pipeline(Input10)
m10.fit(X9, y)
```

```python
yhat_m10 = m10.predict(X9)
yhat_m10
```

```python
ax1 = sns.kdeplot(data=df, x="price")
sns.kdeplot(x=yhat_m10, ax=ax1)
```

```python
ax1 = sns.kdeplot(data=df, x="price")
sns.kdeplot(x=yhat_m9, ax=ax1)
sns.kdeplot(x=yhat_m10, ax=ax1)
```

```python
m9.score(X9, y)
```

```python
m10.score(X9, y)
```

```python
# thử bậc 16 => lệch khá nhiều ko khít
Input16 = [
    ("ss", StandardScaler()),
    ("pf16", PolynomialFeatures(degree=16, include_bias=False)),
    ("m16", LinearRegression())
]

m10b = Pipeline(Input16)
m10b.fit(X9, y)
m10b.score(X9, y)
```

```python
yhat_m10b = m10b.predict(X9)

ax1 = sns.kdeplot(data=df, x="price")
sns.kdeplot(x=yhat_m10b, ax=ax1)
```
